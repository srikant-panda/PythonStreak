```python

from typing import Any


def find_unique_pair(l : list[int],k : int) -> None | str:

    pairs: set[tuple] = set()
    for m,i in enumerate(l):
        for j in l[m+1:]:
            if i + j == k:
                pairs.add((min(i,j),max(i,j)))
    if pairs:
        for i in pairs:
            if i == None:
                return
            print(i)
    else:
        return "No pair available."
find_unique_pair([1,3,2,5,2],4)
```
