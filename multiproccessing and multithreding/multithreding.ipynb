Multithreading in Python

    Multithreading → Multiple threads sharing the same memory space (concurrency).


1. The problem: sequential execution

```python
import time


def task(name):
    print(f"Starting task {name}")
    time.sleep(2)
    print(f"Completed task {name}")


start = time.time()

task("A")
task("B")

end = time.time()
print(f"Total time: {end - start:.2f} seconds")
```

op :- 
Total time: 4.00 seconds

this take 2 second for each task so it take 4 second which can make our code slow in very high cases like if this is an user reques so the other user must wait for complemtion of this user request.

2. Using multithreding

```python
#first import the tread module
from threading import Thread
import time # to calculate the time


#Define the task
def task(name):
    print(f"Starting task {name}")
    time.sleep(2)
    print(f"Completed task {name}")


start = time.time()

# Create Threads
t1 = Thread(target=task,args=("Task-1",))
t2 = Thread(target=task,args=("Task-2",))


# Start both
t1.start()
t2.start()

# wait for both to finish
t1.join()
t2.join()

end = time.time()

print(f"Total time taken : {end - start:.2f} seconds.")
```
