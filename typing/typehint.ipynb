3.Function with multiple type

```python
from typing import Union # Union give the power too use the multiple type to a parameter in function . sysntax : Uninon[...]

def  process_data(data : Union[int,str,float])-> str:
    return f"Proccessed data : {data}"

print(type(process_data(1)))
print(process_data("Srikant"))
print(process_data(1.2))
```

another method of union

```python
def  process_data(data : int | str | float)-> str:  # here "|" means as like as Union
    return f"Proccessed data : {data}"

print(type(process_data(1)))
print(process_data("Srikant"))
print(process_data(1.2))
```

Because typing List and list is not same and same condition fo rest also

```python
from typing import List,Tuple,Dict,Union,Any

def display_item(items : List[str]):
    for i in items:
        print(i)

def get_user() -> Tuple[int,str]:
    return 101,"Srikant"

def get_scores(**data) -> Dict[Any,Any]:
    return data

display_item(['Book','Food',"Computer"])
print(get_user())
d = {
    "name" : "Srikant Panda",
    "id" : 101
}
print(get_scores(**d))
```

5. Optional parameter  sysntax: Optional[str]

        -> optional says you can give me a or you can give None effect

```python
from typing import Optional

def greet(name : Optional[str]) -> None:
    if name:
        print("Welcome",name)
    else:
        print("Welcome user!")

greet(None)
greet('Srikant')
```

6. Type alices

    ----> Alices means creating a datatype by combing multiple datatype.

```python
from typing import Dict,List

# Creating  a custom datatype
UserData = Dict[str,int]

def show_user(data: UserData) -> None:
    for key,value in data.items():
        print(key,"=",value)

payload = {
    "Srikant Panda": 10,
    "Purna" : 12
}
show_user(payload)
```

7. Generic type Hints

```python
from typing import  TypeVar

T = TypeVar("T")  # Thi declares a generic type variable  Which can contain any type of data

def get_first_user(data : List[T]) -> T:
    return data[0]

print(get_first_user([1,2,3,4]))  # o/p - 1
print(get_first_user(['Srikant',"Chandan","Mihir"]))  # o/p - Srikant
```

8.Callable(Function as arguments)

```python
def main(*a)-> int:
    return add_number(*a)
def add_number(*a) -> int:
    total =0 
    for i in a:
        total += i
    return total

main(1,2,3,5)
```

9. Using Callable

```python
from typing import Callable


def main(a: int,b:int,operation: Callable[[int,int],int]) -> int:
    return operation(a,b)

def add(a: int,b : int) -> int:
    return a+b
def sub(a: int,b : int) -> int:
    return a-b
def mul(a: int,b : int) -> int:
    return a*b


print(main(1,2,add))
print(main(1,2,sub))
print(main(1,2,mul))
```

10 Any or NoReturn

```python
from typing import Any, NoReturn


def print_value(value: Any) -> None:
    print("Value:", value)


def fatal_error(message: str) -> NoReturn:
    raise SystemExit(message)


print_value(123)
print_value("Hello")
fatal_error("Something went wrong")
```

11. Type Hinting for classes and Methods

```python
class Main:
    def __init__(self,**kargs:str) -> None:
        self.data = kargs
    def show_details(self) -> None:
        print(self.data)

data = {
    "name" : "Srikant Panda",
    "email" : "srikantap571@gmail.com"
}

obj = Main(**data).show_details()
```

12. Function Annotations (Before Type Hints Era)

```python
def multiply(a: "int value", b: "int value") -> "int result":
    return a * b


print(multiply(3, 4))
print(multiply.__annotations__)  # view stored annotations
```

13. Using Literals

```python
from typing import Literal

def status_check(status: Literal['todo','pending','in-progress']) -> None:
    print(status)

status_check('todo')
status_check('pending')
status_check('completed')
```
